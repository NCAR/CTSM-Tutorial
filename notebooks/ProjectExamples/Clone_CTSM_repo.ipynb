# Clone the CTSM repository
[CTSM is on github](https://github.com/ESCOMP/ctsm) and freely available for non-commercial use!
To get started using CTSM you have to:
1. Clone the repository
2. Get necessary submodules (e.g. FATES, mosart, cime, etc.)
---
Porting the whole CESM / CTSM infrastructre and data needed to run the model so you can run the code on a new machine is a bit more challenging, but we've taken care of that if you're running on an NCAR supported system.  There alre also resources for [porting on the CESM bulletin board](https://bb.cgd.ucar.edu/cesm/tags/porting/). 

The example below will clone ctsm into your home directory

```bash
git clone https://github.com/ESCOMP/ctsm ~/ctsm
cd ~/ctsm
./bin/git-fleximod update
```

What version of CTSM do you have?  You can check this with `git describe`

```bash
git describe
```

The example above gets you the latest development code of CTSM!

These development tags are just that, code that's under development, and the CTSM code is changing all the time.  These software engineering advancements  still allow you to do science with the code. By getting the lates code base, you know you're getting the latest and greatest CTSM code, but it may also have bugs, issues, or problems that we may (or may not) know about.

We also have major version and minor version release tags that you can checkout on the CTSM github page.  These release versions of the code also have papers (for major versions) and dois (major and minor versions) which are helpful to reference when you're ready to publish your work.
